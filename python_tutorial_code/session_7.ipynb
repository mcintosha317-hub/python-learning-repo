```python
# if -> does the thing if its true (can add anywhere)
# elif (else if) -> does the thing if it's true buit the first bit wasn't
# else -> does the thing if the if wasn't true

#spacing / indents important to get right

names_list = ['Will', 'Clare', 'Marcel', 'Joe', 'Alistair']

for name in names_list:
    if name == 'Marcel':
        print("Woof!")
    elif name == 'Will':
        print("Hi, my name is Will!")
    else:
        print(f'Hello, my name is {name}!')
    if name == 'Alistair':
        print("I am Will's boss")
```

```python
evens = list(range(2,13,2)) # start,stop,step (start at 2, stope at 13 and give every second one)
odds = list(range(1,13,2))

multiply_list = []

for even in evens:
    for odd in odds:
        #print(even)
        #print(hi)
        result = even * odd
        statement = f'{even} times {odd} is {result}'
        #print(statement)
        multiply_list.append(statement)

print(multiply_list)
```

```python
# print(6 % 5 == 0)
# print(6 % 1 == 0)

# fizzbang
# take all the numbers from 1-100, 
# print them, BUT if the number is divisible by 3, print fizz,
# if it's divisible by 5 print bang
# BUT if it's divisible by 3 & 5 print fizzbang

nums = list(range(1,101))

for num in nums:
    if (num % 3 == 0) & (num % 5 == 0):
        print('fizzbang') 
    elif num % 3 == 0:
        print('fizz')
    elif num % 5 == 0: 
        print('bang')
    else:
        print(num)

 
```
